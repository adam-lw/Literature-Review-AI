```python
from papery.core.api.core import api_call

data = await api_call(
    params={"query": "machine learning medical diagnosis", "fields": "title,abstract"},
    api_id="semantic_scholar_api",
    endpoint="graph/v1/paper/search/bulk/",
)

import pandas as pd

results = pd.DataFrame(data.get("data", None))
display(results)
```

```python
from papery.core.db import save_table_async, load_table

await save_table_async(
    results, table_path="public.test_papers_basic", if_exists="append"
)
```
