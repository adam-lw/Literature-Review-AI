```python
from papery.core.db import load_table, save_table
import pandas as pd


def deduplicate_data(table_uri: str, key: str) -> pd.DataFrame:
    """
    Deduplicates data in a table based on a specified key.

    Args:
        table_uri (str): The URI of the table to deduplicate.
        key (str): The key to use for deduplication.

    Returns:
        DataFrame: A deduplicated DataFrame.
    """
    # Load the table data
    table_data = load_table(table_uri)

    print(f"Original data: {len(table_data)}")

    df_deduped = table_data.drop_duplicates(subset=[key])

    print(len(df_deduped))
    return df_deduped


papers_uri = "test.test_papers_dataset"
df = deduplicate_data(papers_uri, "paperId")
save_table(df, papers_uri, if_exists="replace")
```
